```python
import pandas as pd
```

```python
# Series or one dim array
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(data)
print(series)
```

```python
# Custom Index
# Specific labels to each element
series1 = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series1)
```

```python
# Dataframe
# Making a table-like structure (Excel Spreadsheet)

new_data = {
    'Name': ['John', 'Paul', 'Ringo', 'George'],
    'Age': [25, 35, 25, 0],
    'City': ['New York', 'Los Angeles', 'Chicago', '0']}
df = pd.DataFrame(new_data)
print(df)
```

```python
print(df.shape)
print(df.columns)
# print(df.info())
# print(df.describe())
```

```python
print(df['Age']) # Selecting age
print(df[['Age', 'Name']]) # Selecting columns
```

```python
filtered_df = df[df['Age'] > 26]
print(filtered_df)
```

```python
df_sorted = df.sort_values(by='Age', ascending=False)
print(df_sorted)
```

```python
print(df.isnull().sum())
```

```python
# Plotting
import matplotlib.pyplot as plt
```

```python
df['Age'].plot(kind='hist', title='Ages')
plt.show()
```
